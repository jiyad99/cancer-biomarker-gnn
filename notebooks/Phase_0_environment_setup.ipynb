## Cell 6 — GPU Verification

Expected:
- CUDA available: **True**
- VRAM: **≥ 3.5 GB** (4 GB card with ~0.5 GB driver overhead)
- A dummy tensor forward pass succeeds

If CUDA is **False**: PyTorch installed the CPU-only build. Re-run Cell 3 with the correct CUDA index URL.

````python
import torch

print(f"PyTorch version      : {torch.__version__}")
print(f"CUDA available       : {torch.cuda.is_available()}")

if not torch.cuda.is_available():
    print()
    print("✗ CUDA not available. Likely causes:")
    print("  1. PyTorch CPU-only build — re-run Cell 3 with the correct --index-url")
    print("  2. NVIDIA driver not installed — check nvidia-smi in a terminal")
    print("  3. CUDA toolkit version mismatch — check nvcc --version")
else:
    device = torch.device('cuda')
    props  = torch.cuda.get_device_properties(0)

    vram_gb = props.total_memory / 1024**3
    print(f"GPU name             : {props.name}")
    print(f"VRAM total           : {vram_gb:.2f} GB")
    print(f"CUDA capability      : {props.major}.{props.minor}")
    print(f"torch.cuda.device    : {torch.cuda.current_device()}")

    # Dummy forward pass — catches driver/library mismatches that survive import
    x = torch.randn(100, 64, device=device)
    y = torch.nn.Linear(64, 1).to(device)(x)
    assert y.shape == (100, 1), "Unexpected output shape"
    del x, y
    torch.cuda.empty_cache()

    if vram_gb < 3.0:
        print(f"\n⚠ Only {vram_gb:.1f} GB VRAM detected — below the 4 GB minimum.")
        print("  Training may OOM. Gradient accumulation (micro-batch=2) will be required.")
    else:
        print(f"\n✓ GPU verified. {vram_gb:.1f} GB VRAM available.")
        print("  Proceed. A VRAM dry-run is scheduled in Phase 8 (before full 5×3 CV).")
````

## Cell 9 — Logging Setup

Creates `src/utils/logger.py` — a single `get_logger(phase_name)` factory used by every notebook.
Logs go to `results/logs/<phase>_<timestamp>.log` AND print to stdout.

Then runs a smoke test to confirm the log file is created.

````python
logger_content = '''
"""GAT-v4 | Logging factory.

Usage in any notebook or script:
    from src.utils.logger import get_logger
    log = get_logger('phase1_data')
    log.info('Starting GDC download')
"""

import logging
from datetime import datetime
from pathlib import Path

# Import LOGS path from config, with fallback for standalone use
try:
    from src.config import LOGS
except ImportError:
    LOGS = Path("results/logs")


def get_logger(phase_name: str, level: int = logging.INFO) -> logging.Logger:
    """Return a logger that writes to both stdout and a timestamped file.

    Args:
        phase_name: Short string identifying the phase, e.g. 'phase1_data'.
        level:      Logging level (default INFO).

    Returns:
        Configured Logger instance.
    """
    LOGS.mkdir(parents=True, exist_ok=True)

    timestamp  = datetime.now().strftime("%Y%m%d_%H%M%S")
    log_file   = LOGS / f"{phase_name}_{timestamp}.log"

    logger = logging.getLogger(phase_name)
    logger.setLevel(level)

    # Avoid duplicate handlers if cell is re-run
    if logger.hasHandlers():
        logger.handlers.clear()

    fmt = logging.Formatter(
        fmt="%(asctime)s | %(levelname)-8s | %(name)s | %(message)s",
        datefmt="%Y-%m-%d %H:%M:%S"
    )

    # File handler
    fh = logging.FileHandler(log_file, encoding="utf-8")
    fh.setLevel(level)
    fh.setFormatter(fmt)
    logger.addHandler(fh)

    # Stream handler (stdout)
    sh = logging.StreamHandler()
    sh.setLevel(level)
    sh.setFormatter(fmt)
    logger.addHandler(sh)

    logger.info(f"Logger initialised. Log file → {log_file}")
    return logger
'''

logger_path = PROJECT_ROOT / 'src' / 'utils' / 'logger.py'
logger_path.write_text(logger_content.strip(), encoding='utf-8')
print(f"Written: {logger_path}")

# Smoke test
importlib.invalidate_caches()
from src.utils.logger import get_logger

log = get_logger('phase0_setup')
log.info('Phase 0 environment setup — smoke test')
log.info(f'Python {sys.version}')

import torch
log.info(f'PyTorch {torch.__version__} | CUDA available: {torch.cuda.is_available()}')

print()
# Confirm log file exists
log_files = list((PROJECT_ROOT / 'results' / 'logs').glob('phase0_setup_*.log'))
if log_files:
    print(f"✓ Log file created: {log_files[-1].name}")
else:
    print("✗ Log file not found — check LOGS path in src/config.py")
````

## Cell 12 — Phase 0 Audit YAML

Records what was installed, when, and on what hardware.
Saved to `audit/phase0_environment.yaml` — commit this file.

A reviewer asking "what GPU did you use?" is answered by this file.

````python
import yaml
import torch
from datetime import datetime
import subprocess, sys, platform

# Collect installed package versions
def get_pkg_version(pkg_name):
    try:
        mod = importlib.import_module(pkg_name)
        return getattr(mod, '__version__', 'unknown')
    except ImportError:
        return 'NOT INSTALLED'

# GPU info
if torch.cuda.is_available():
    props = torch.cuda.get_device_properties(0)
    gpu_info = {
        'name': props.name,
        'vram_gb': round(props.total_memory / 1024**3, 2),
        'cuda_capability': f"{props.major}.{props.minor}",
        'cuda_version': torch.version.cuda,
    }
else:
    gpu_info = {'name': 'NOT AVAILABLE', 'vram_gb': 0}

audit = {
    'phase': 'Phase 0 — Environment Setup',
    'timestamp': datetime.now().isoformat(),
    'system': {
        'os': platform.platform(),
        'python': sys.version,
        'architecture': platform.machine(),
    },
    'gpu': gpu_info,
    'packages': {
        'torch': get_pkg_version('torch'),
        'torch_geometric': get_pkg_version('torch_geometric'),
        'captum': get_pkg_version('captum'),
        'numpy': get_pkg_version('numpy'),
        'pandas': get_pkg_version('pandas'),
        'scipy': get_pkg_version('scipy'),
        'lifelines': get_pkg_version('lifelines'),
        'scikit_survival': get_pkg_version('sksurv'),
        'pycox': get_pkg_version('pycox'),
        'networkx': get_pkg_version('networkx'),
        'mygene': get_pkg_version('mygene'),
        'gseapy': get_pkg_version('gseapy'),
    },
    'seed_registry': {
        'master_seed': 42,
        'cv_repeat_seeds': [42, 123, 456],
        'n_cv_folds': 5,
        'n_bootstrap': 1000,
        'note': 'FROZEN — do not change after Phase 1 data acquisition begins',
    },
    'directory_root': str(PROJECT_ROOT),
    'status': 'COMPLETE',
}

audit_path = PROJECT_ROOT / 'audit' / 'phase0_environment.yaml'
with open(audit_path, 'w', encoding='utf-8') as f:
    yaml.dump(audit, f, default_flow_style=False, sort_keys=False, allow_unicode=True)

print(f"Written: {audit_path}")
print()
print(yaml.dump(audit, default_flow_style=False, sort_keys=False))
````
